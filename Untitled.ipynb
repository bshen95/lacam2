```python
import matplotlib.pyplot as plt
import numpy as np
def read_map_file(filepath):
    with open(filepath, 'r') as file:
        lines = file.readlines()

    # Skip the first four lines (metadata)
    map_data = lines[4:]

    return [line.strip() for line in map_data]

def visualize_map(map_data):
    height = len(map_data)
    width = len(map_data[0])

    # Create a numpy array for the map
    map_array = np.zeros((height, width))

    for y, row in enumerate(map_data):
        for x, cell in enumerate(row):
            if cell == '@':
                map_array[y, x] = 1  # Obstacles

    plt.imshow(map_array, cmap='Greys', origin='upper')
    plt.title('Map Visualization')
    plt.show()
```
